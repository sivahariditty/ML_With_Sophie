# 1. Generate and Load Data
Let's start by importing our libraries and generating a simple 1D dataset for regression.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.pipeline import make_pipeline
from sklearn.datasets import make_moons

# Generate synthetic regression data
np.random.seed(42)
X = np.sort(np.random.rand(20) * 10).reshape(-1, 1)
y = 2 * X.ravel() + 1 + np.random.randn(20) * 3
```

# 2. Inspect the Data
Let's check how many data points we have generated.

```python
print(f"Number of data points: {len(X)}")
```

# 3. Plot the Raw Data
Visualizing the dataset before applying any models.

```python
plt.scatter(X, y, color='blue', label='Data Points')
plt.title('Raw Training Data')
plt.xlabel('X')
plt.ylabel('y')
plt.legend()
plt.show()
```

# 4. Fit a Simple Linear Model
Now we will fit a standard linear regression model to our data.

```python
model = LinearRegression()
model.fit(X, y)
print("Linear Regression model fitted successfully.")
```

# 5. Plot the Linear Model
Let's see how our basic linear model fits the data. This represents a simple model that might be underfitting slightly.

```python
X_plot = np.linspace(0, 10, 100).reshape(-1, 1)

plt.scatter(X, y, color='blue', label='Data Points')
plt.plot(X_plot, model.predict(X_plot), color='red', label='Linear Fit')
plt.title('Basic Linear Regression')
plt.legend()
plt.show()
```

# 6. Increase Complexity: Polynomial Degree 3
We increase the model complexity by adding polynomial features up to degree 3.

```python
poly3 = make_pipeline(PolynomialFeatures(degree=3), LinearRegression())
poly3.fit(X, y)

plt.scatter(X, y, color='blue', label='Data Points')
plt.plot(X_plot, poly3.predict(X_plot), color='green', label='Degree 3 Fit')
plt.title('Polynomial Regression (Degree 3)')
plt.ylim(-5, 30)
plt.legend()
plt.show()
```

# 7. Increase Complexity: Polynomial Degree 7
Increasing the complexity further to degree 7. The curve will start to wiggle more to hit the points.

```python
poly7 = make_pipeline(PolynomialFeatures(degree=7), LinearRegression())
poly7.fit(X, y)

plt.scatter(X, y, color='blue', label='Data Points')
plt.plot(X_plot, poly7.predict(X_plot), color='orange', label='Degree 7 Fit')
plt.title('Polynomial Regression (Degree 7)')
plt.ylim(-5, 30)
plt.legend()
plt.show()
```

# 8. Maximum Complexity: Polynomial Degree 15 (Overfitting)
At degree 15, the model has almost enough capacity to perfectly memorize the training data, leading to severe overfitting. It catches the noise, not the underlying trend.

```python
poly15 = make_pipeline(PolynomialFeatures(degree=15), LinearRegression())
poly15.fit(X, y)

plt.scatter(X, y, color='blue', label='Data Points')
plt.plot(X_plot, poly15.predict(X_plot), color='purple', label='Degree 15 Fit')
plt.title('Polynomial Regression (Degree 15) - OVERFITTING')
plt.ylim(-5, 30)
plt.legend()
plt.show()
```

# 9. Classification: Generate Data
Let's switch to a classification task. We will generate a non-linear dataset.

```python
X_cls, y_cls = make_moons(n_samples=100, noise=0.2, random_state=42)
print(f"Generated classification dataset with {len(X_cls)} samples.")
```

# 10. Define Decision Boundary Plotter
A helper function to visualize how classification models separate the two classes.

```python
def plot_decision_boundary(model, X, y, title):
    h = .02
    x_min, x_max = X[:, 0].min() - 0.5, X[:, 0].max() + 0.5
    y_min, y_max = X[:, 1].min() - 0.5, X[:, 1].max() + 0.5
    xx, yy = np.meshgrid(np.arange(x_min, x_max, h), np.arange(y_min, y_max, h))
    
    Z = model.predict(np.c_[xx.ravel(), yy.ravel()])
    Z = Z.reshape(xx.shape)
    
    plt.contourf(xx, yy, Z, alpha=0.3, cmap=plt.cm.coolwarm)
    plt.scatter(X[:, 0], X[:, 1], c=y, edgecolors='k', cmap=plt.cm.coolwarm)
    plt.title(title)
    plt.show()
```

# 11. Plot Raw Classification Data
Visualizing the two classes before fitting any models.

```python
plt.scatter(X_cls[:, 0], X_cls[:, 1], c=y_cls, edgecolors='k', cmap=plt.cm.coolwarm)
plt.title('Raw Classification Data (Make Moons)')
plt.show()
```

# 12. Fit Simple Logistic Regression
A simple linear classifier that will struggle to separate the non-linear moons properly.

```python
clf = LogisticRegression()
clf.fit(X_cls, y_cls)
plot_decision_boundary(clf, X_cls, y_cls, 'Simple Logistic Regression (Underfitting)')
```

# 13. Complex Classification (Overfitting)
We dramatically increase the complexity by projecting features into a 15th-degree polynomial space.

```python
complex_clf = make_pipeline(PolynomialFeatures(degree=15), LogisticRegression(max_iter=10000, C=1000))
complex_clf.fit(X_cls, y_cls)
plot_decision_boundary(complex_clf, X_cls, y_cls, 'High Degree Polynomial (Overfitting)')
```

# 14. Add More Data
What happens if we take the exact same extremely complex model (Degree 15), but give it vastly more data?

```python
X_cls_large, y_cls_large = make_moons(n_samples=2000, noise=0.2, random_state=42)
print(f"Generated LARGE classification dataset with {len(X_cls_large)} samples.")
```

# 15. Why Does More Data Reduce Overfitting?
Overfitting occurs when a complex model memorizes the random noise and outliers in a dataset rather than learning the actual underlying pattern. When you only have 100 data points, a 15th-degree polynomial has enough mathematical flexibility to effectively 'connect the dots' and twist its decision boundary around every single noisy anomaly.

When you scale up to 2,000 data points, the random noise cancels itself out and the true 'signal' (the shape of the moons) becomes dominant. Even though the model is still highly complex, it no longer possesses enough capacity to memorize thousands of random noise points. It is therefore forced to fit the general, underlying shape of the distribution instead.

```python
complex_clf_large = make_pipeline(PolynomialFeatures(degree=15), LogisticRegression(max_iter=10000, C=1000))
complex_clf_large.fit(X_cls_large, y_cls_large)
plot_decision_boundary(complex_clf_large, X_cls_large, y_cls_large, 'Complex Model + Large Data (Reduced Overfitting)')
```
